```python
# My first calculator in Python

print("=== WELCOME TO YOUR PYTHON CALCULATOR ===")

# 1. Ask the user for two numbers
# We use float() so the calculator can handle decimal numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# 2. Show the options menu
print("\nWhat operation would you like to perform?")
print("1. Add (+)")
print("2. Subtract (-)")
print("3. Multiply (*)")
print("4. Divide (/)")
print("5. Power (**)")

choice = input("Choose an option (1, 2, 3, 4, or 5): ")

# 3. Perform the operation based on the user's choice
if choice == '1':
    result = num1 + num2
    print(f"\nResult: {num1} + {num2} = {result}")

elif choice == '2':
    result = num1 - num2
    print(f"\nResult: {num1} - {num2} = {result}")

elif choice == '3':
    result = num1 * num2
    print(f"\nResult: {num1} * {num2} = {result}")

elif choice == '4':
    # We check if num2 is not zero to prevent a crash
    if num2 != 0:
        result = num1 / num2
        print(f"\nResult: {num1} / {num2} = {result}")
elif choice == '5':
    if num2 != 0:
        result = num1 ** num2
        print(f"\nResult: {num1} ** {num2} = {result}")
    else:
        print("\nError: Cannot divide by zero.")

else:
    print("\nInvalid option. Please run the program again.")
```
